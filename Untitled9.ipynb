```python

# ============================================================
# FACTORY PRODUCTION PLANNER (A7)
# B.Tech Mathematics Mini Project
# Concept: Consistency, Rank Analysis & Practical Interpretation
# Environment: Google Colab + Python
# ============================================================


# ============================================================
# CELL 1: IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display, Markdown


# ============================================================
# CELL 2: PROJECT TITLE
# ============================================================

display(Markdown("""
# 🏭 FACTORY PRODUCTION PLANNER (A7)

### B.Tech Mathematics Mini Project

**Mathematical Concepts:**
- System of Linear Equations
- Consistency of equations
- Rank of a matrix
- Gaussian elimination
- Practical bottleneck analysis

**Real-world context:**
Manufacturing plants use mathematical models to decide
how many units of different products can be produced
with limited machine capacity.
"""))


# ============================================================
# CELL 3: INPUT CONFIGURATION
# ============================================================
# You can change these values to test different factories.
#
# Products:
#   x = Product 1 quantity
#   y = Product 2 quantity
#   z = Product 3 quantity
#
# Machines:
#   M1, M2, M3
#
# Each row represents one machine.
# Each column represents one product.
#
# Example:
# M1: 2x + 1y + 1z = 100
# M2: 1x + 2y + 1z = 100
# M3: 1x + 1y + 2z = 100


# Product names
products = ["Product A", "Product B", "Product C"]

# Machine names
machines = ["Machine 1", "Machine 2", "Machine 3"]


# Time required by each product on each machine
#
#              Product A   Product B   Product C
machine_times = np.array([
    [2,           1,          1],   # Machine 1
    [1,           2,          1],   # Machine 2
    [1,           1,          2]    # Machine 3
], dtype=float)


# Available working hours for each machine
machine_capacity = np.array([
    100,   # Machine 1
    100,   # Machine 2
    100    # Machine 3
], dtype=float)


# ============================================================
# CELL 4: INPUT VALIDATION
# ============================================================

if machine_times.shape != (3, 3):
    raise ValueError(
        "machine_times must contain exactly 3 machines and 3 products."
    )

if len(machine_capacity) != 3:
    raise ValueError(
        "machine_capacity must contain exactly 3 machine capacities."
    )

if np.any(machine_times < 0):
    raise ValueError(
        "Processing times cannot be negative."
    )

if np.any(machine_capacity < 0):
    raise ValueError(
        "Machine capacities cannot be negative."
    )


display(Markdown("## ⚙️ Factory Input Configuration"))

print("Products:")
for i, p in enumerate(products):
    print(f"  {i+1}. {p}")

print("\nMachines:")
for i, m in enumerate(machines):
    print(
        f"  {m}: {machine_capacity[i]:.2f} available hours"
    )

print("\nProcessing Time Matrix:")
print(machine_times)


# ============================================================
# CELL 5: DISPLAY MATHEMATICAL MODEL
# ============================================================

display(Markdown("## 📐 Mathematical Model"))

print("Let:")
print("  x = quantity of Product A")
print("  y = quantity of Product B")
print("  z = quantity of Product C")

print("\nThe machine constraints are:")

for i in range(3):

    equation = ""

    for j in range(3):

        coefficient = machine_times[i][j]

        if j == 0:
            variable = "x"
        elif j == 1:
            variable = "y"
        else:
            variable = "z"

        if j == 0:
            equation += f"{coefficient:g}{variable}"
        else:
            equation += f" + {coefficient:g}{variable}"

    equation += f" = {machine_capacity[i]:g}"

    print(f"{machines[i]}: {equation}")


# ============================================================
# CELL 6: CREATE MATRIX REPRESENTATION
# ============================================================

A = sp.Matrix(machine_times.astype(int))
b = sp.Matrix(machine_capacity.astype(int))

augmented_matrix = A.row_join(b)

display(Markdown("## 🔢 Matrix Representation"))

print("Coefficient Matrix A:")
display(A)

print("\nCapacity Vector b:")
display(b)

print("\nAugmented Matrix [A | b]:")
display(augmented_matrix)


# ============================================================
# CELL 7: RANK ANALYSIS
# ============================================================

rank_A = A.rank()
rank_augmented = augmented_matrix.rank()

display(Markdown("## 📊 Consistency & Rank Analysis"))

print(f"Rank(A)       = {rank_A}")
print(f"Rank([A | b]) = {rank_augmented}")

print("\nConsistency Rule:")
print("A system is consistent when:")
print("Rank(A) = Rank([A | b])")

print("\nResult:")

if rank_A == rank_augmented:
    print("✅ The system is CONSISTENT.")

    if rank_A == 3:
        print("✅ The system has a UNIQUE solution.")
    else:
        print("ℹ️ The system has infinitely many solutions.")

else:
    print("❌ The system is INCONSISTENT.")
    print("The machine constraints cannot all be satisfied simultaneously.")


# ============================================================
# CELL 8: GAUSSIAN ELIMINATION / RREF
# ============================================================

display(Markdown("## 🔄 Gaussian Elimination"))

print("Reduced Row Echelon Form of [A | b]:")

rref_matrix, pivot_columns = augmented_matrix.rref()

display(rref_matrix)

print("Pivot columns:", pivot_columns)


# ============================================================
# CELL 9: NUMERICAL SOLUTION
# ============================================================

display(Markdown("## 🧮 Production Quantity Calculation"))

solution_exists = False
solution = None

if rank_A == rank_augmented:

    if rank_A == 3:

        # Solve the square system
        solution = A.LUsolve(b)
        solution_exists = True

        print("Unique production solution:")

        for i, value in enumerate(solution):
            print(
                f"{products[i]} = {float(value):.2f} units"
            )

    else:
        print(
            "The system is consistent but does not have a unique solution."
        )

else:

    print(
        "No exact production quantities satisfy all three "
        "machine equations simultaneously."
    )


# ============================================================
# CELL 10: VERIFY THE SOLUTION
# ============================================================

if solution_exists:

    display(Markdown("## ✅ Solution Verification"))

    verification = A * solution

    print("Calculated machine usage:")

    for i in range(3):

        calculated = float(verification[i])
        available = float(b[i])

        print(
            f"{machines[i]}: "
            f"{calculated:.2f} / {available:.2f} hours"
        )

    print("\nVerification:")
    print("A × solution = b")

    display(verification)


# ============================================================
# CELL 11: PRODUCTION TABLE
# ============================================================

if solution_exists:

    display(Markdown("## 📋 Production Plan"))

    print(
        f"{'Product':<15}"
        f"{'Quantity':>15}"
    )

    print("-" * 30)

    for i in range(3):

        print(
            f"{products[i]:<15}"
            f"{float(solution[i]):>15.2f}"
        )


# ============================================================
# CELL 12: MACHINE UTILIZATION
# ============================================================

if solution_exists:

    usage = np.array(
        [float(value) for value in verification]
    )

    utilization = (
        usage / machine_capacity
    ) * 100

    display(Markdown("## ⚙️ Machine Utilization"))

    print(
        f"{'Machine':<15}"
        f"{'Used Hours':>15}"
        f"{'Capacity':>15}"
        f"{'Utilization':>15}"
    )

    print("-" * 60)

    for i in range(3):

        print(
            f"{machines[i]:<15}"
            f"{usage[i]:>15.2f}"
            f"{machine_capacity[i]:>15.2f}"
            f"{utilization[i]:>14.2f}%"
        )


# ============================================================
# CELL 13: BOTTLENECK ANALYSIS
# ============================================================
#
# Assumption:
# A bottleneck is the machine with the highest utilization
# relative to its available capacity.
#
# This definition is only used for the practical interpretation.
# It does NOT change the mathematical solution.


display(Markdown("## 🚧 Bottleneck Analysis"))

if solution_exists:

    bottleneck_index = np.argmax(utilization)

    print(
        f"Potential bottleneck: "
        f"{machines[bottleneck_index]}"
    )

    print(
        f"Utilization: "
        f"{utilization[bottleneck_index]:.2f}%"
    )

    if utilization[bottleneck_index] >= 90:
        print(
            "⚠️ This machine is highly utilized and may limit "
            "future production."
        )

    elif utilization[bottleneck_index] >= 75:
        print(
            "⚠️ This machine has relatively high utilization."
        )

    else:
        print(
            "ℹ️ No machine is heavily utilized in this production plan."
        )

else:

    print(
        "Exact bottleneck analysis cannot be calculated from "
        "the inconsistent equality system."
    )


# ============================================================
# CELL 14: CHART 1 — PRODUCTION QUANTITIES
# ============================================================

if solution_exists:

    plt.figure(figsize=(9, 5))

    quantities = [
        float(value)
        for value in solution
    ]

    bars = plt.bar(
        products,
        quantities
    )

    plt.title(
        "Factory Production Quantities",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel("Products")
    plt.ylabel("Number of Units")

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    for bar, value in zip(bars, quantities):

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f"{value:.2f}",
            ha="center",
            va="bottom"
        )

    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 15: CHART 2 — MACHINE UTILIZATION
# ============================================================

if solution_exists:

    plt.figure(figsize=(9, 5))

    bars = plt.bar(
        machines,
        utilization
    )

    plt.axhline(
        100,
        linestyle="--",
        linewidth=2,
        label="100% Capacity"
    )

    plt.title(
        "Machine Utilization",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel("Machines")
    plt.ylabel("Utilization (%)")

    plt.ylim(
        0,
        max(110, max(utilization) + 10)
    )

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    for bar, value in zip(bars, utilization):

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f"{value:.1f}%",
            ha="center",
            va="bottom"
        )

    plt.legend()
    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 16: CHART 3 — MACHINE HOURS USED VS AVAILABLE
# ============================================================

if solution_exists:

    x_pos = np.arange(len(machines))

    width = 0.35

    plt.figure(figsize=(10, 5))

    plt.bar(
        x_pos - width / 2,
        usage,
        width,
        label="Used Hours"
    )

    plt.bar(
        x_pos + width / 2,
        machine_capacity,
        width,
        label="Available Hours"
    )

    plt.xticks(
        x_pos,
        machines
    )

    plt.ylabel("Hours")

    plt.title(
        "Machine Hours: Used vs Available",
        fontsize=16,
        fontweight="bold"
    )

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.legend()

    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 17: HEATMAP — PROCESSING TIME MATRIX
# ============================================================

plt.figure(figsize=(8, 5))

plt.imshow(
    machine_times,
    aspect="auto"
)

plt.colorbar(
    label="Processing Time"
)

plt.xticks(
    range(3),
    products
)

plt.yticks(
    range(3),
    machines
)

plt.title(
    "Processing Time Matrix",
    fontsize=16,
    fontweight="bold"
)

for i in range(3):

    for j in range(3):

        plt.text(
            j,
            i,
            f"{machine_times[i, j]:g}",
            ha="center",
            va="center"
        )

plt.xlabel("Products")
plt.ylabel("Machines")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 18: FINAL PROJECT SUMMARY
# ============================================================

display(Markdown("## 📝 Final Project Summary"))

print("PROJECT: Factory Production Planner (A7)")
print("=" * 50)

print("\nMathematical Model:")
print("A × X = B")

print("\nWhere:")
print("A = Machine/Product processing-time matrix")
print("X = Unknown production quantities")
print("B = Available machine capacity vector")

print("\nMathematical Analysis:")
print(f"Rank(A) = {rank_A}")
print(f"Rank([A|B]) = {rank_augmented}")

if rank_A == rank_augmented:

    print("System Status: CONSISTENT")

    if solution_exists:
        print("Solution Type: UNIQUE SOLUTION")

        print("\nRecommended Production Quantities:")

        for i in range(3):

            print(
                f"  {products[i]}: "
                f"{float(solution[i]):.2f} units"
            )

        print(
            f"\nPotential Bottleneck: "
            f"{machines[bottleneck_index]}"
        )

else:

    print("System Status: INCONSISTENT")
    print
```
